```python
import numpy as np
import pandas as pd
import pyarrow
import os
```

```python

# Set random seed for consistency
np.random.seed(42)
n_samples = 1200

# Generating Features
distance = np.random.uniform(10, 500, n_samples)          # Distance in miles
traffic_delay = np.random.exponential(2, n_samples)       # Traffic delay in hours
weather_delay = distance * 0.01 + np.random.normal(1, 0.5, n_samples) # Strongly correlated with distance!
```

```python


# Categorical Region Factor
regions = np.random.choice(['East_Hub', 'West_Hub', 'South_Hub'], size=n_samples)

# Create DataFrame
df = pd.DataFrame({
    'Distance_Miles': distance,
    'Traffic_Delay_Hrs': traffic_delay,
    'Weather_Delay_Hrs': weather_delay,
    'Warehouse_Region': regions
})

# True physical formula for Delivery Time + random nature noise
df['Delivery_Time_Hrs'] = (
    5.0 + 
    (df['Distance_Miles'] * 0.15) + 
    (df['Traffic_Delay_Hrs'] * 1.8) + 
    (df['Warehouse_Region'] == 'South_Hub') * 4.5 + 
    np.random.normal(0, 1, n_samples)
)
```
